```python
from pathlib import Path
import numpy as np
import pandas as pd
import plotly.graph_objects as go
import plotly.express as px

import sys
import warnings
from pathlib import Path

import numpy as np
import pandas as pd

pd.set_option("display.max_columns", 220)
pd.set_option("display.max_rows", 300)

CURRENT_DIR = Path.cwd().resolve()

if (CURRENT_DIR / "src").exists():
    PROJECT_ROOT = CURRENT_DIR
elif (CURRENT_DIR.parent / "src").exists():
    PROJECT_ROOT = CURRENT_DIR.parent
else:
    raise RuntimeError(
        "Could not find project root. Launch the notebook from the project "
        "root or from the notebooks/ folder."
    )

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print("PROJECT_ROOT:", PROJECT_ROOT)
PRESENTATION_DIR = PROJECT_ROOT / "results" / "presentation_AG"
PRESENTATION_DIR.mkdir(parents=True, exist_ok=True)


from src.io.database_h5 import load_nir_uco_h5
from src.utils import save_parquet, load_parquet

from src.spectra.preprocessing_configs import normalize_preprocessing_configs
from src.spectra.band_selection import (
    select_wavelength_range_from_database,
    wavelength_selection_summary,
)

from src.workflows.simca_pixel_grid import (
    run_single_simca_pixel_projection,
)
from src.workflows.simca_cv_calibration import (
    run_simca_empirical_rule_grid,
)
from src.decision.metrics import (
    binary_detection_metrics,
    metrics_by_group,
    summarize_pixel_errors_by_image,
)
from src.decision.uncertainty import (
    add_three_way_object_decision,
    summarize_three_way_decision,
)
from src.decision.border import (
    border_width_object_threshold_grid,
    summarize_pixel_errors_by_border_zone,
)

from src.visualization.plot_generic import (
    plot_bar_values,
    plot_counts_by_group,
)
from src.visualization.plot_decision import (
    plot_pixel_error_overlay,
    plot_pixel_fp_fn_overlay,
)

%load_ext autoreload
%autoreload 2

def format_for_ppt(fig, title=None, width=1200, height=750):
    """
    Harmonise les figures pour insertion dans PowerPoint.
    """
    fig.update_layout(
        template="plotly_white",
        width=width,
        height=height,
        title=dict(
            text=title if title is not None else fig.layout.title.text,
            x=0.5,
            xanchor="center",
            font=dict(size=30),
        ),
        font=dict(size=22),
        margin=dict(l=70, r=50, t=90, b=70),
        legend=dict(
            font=dict(size=18),
            orientation="h",
            yanchor="bottom",
            y=-0.25,
            xanchor="center",
            x=0.5,
        ),
    )
    fig.update_xaxes(title_font=dict(size=22), tickfont=dict(size=18))
    fig.update_yaxes(title_font=dict(size=22), tickfont=dict(size=18))
    return fig


def export_for_ppt(fig, name, width=1200, height=750, scale=2):
    """
    Exporte en PNG pour PowerPoint + HTML interactif de secours.
    Nécessite kaleido pour le PNG :
        pip install -U kaleido
    """
    png_path = PRESENTATION_DIR / f"{name}.png"
    html_path = PRESENTATION_DIR / f"{name}.html"

    fig.update_layout(width=width, height=height)

    try:
        fig.write_image(str(png_path), scale=scale)
        print(f"[OK] PNG saved: {png_path}")
    except Exception as exc:
        print("[WARNING] PNG export failed. Install kaleido with: pip install -U kaleido")
        print("Error:", repr(exc))

    fig.write_html(str(html_path))
    print(f"[OK] HTML saved: {html_path}")
```

```python
from scipy.optimize import linear_sum_assignment


def clean_image_key(key):
    key = str(key).strip().lower()
    if key.endswith("_sb"):
        key = key[:-3]
    return key


def get_objects_from_image(object_db, image_key):
    image_key = clean_image_key(image_key)

    objects = {
        oid: obj
        for oid, obj in object_db.items()
        if clean_image_key(obj.get("source_clean_key")) == image_key
    }

    if len(objects) == 0:
        raise ValueError(f"No object found for image_key={image_key!r}")

    return objects


def get_object_global_mask(obj, image_shape):
    """
    Retourne le masque global d'un objet.
    Si mask_global n'existe pas, il est reconstruit à partir de bbox + mask.
    """
    if "mask_global" in obj and obj["mask_global"] is not None:
        return np.asarray(obj["mask_global"]).astype(bool)

    if "bbox" not in obj or "mask" not in obj:
        raise KeyError("Object must contain either mask_global or bbox + mask.")

    min_row, min_col, max_row, max_col = map(int, obj["bbox"])
    mask_crop = np.asarray(obj["mask"]).astype(bool)

    mask_global = np.zeros(image_shape, dtype=bool)
    mask_global[min_row:max_row, min_col:max_col] = mask_crop

    return mask_global


def centroid_distance(obj_a, obj_b):
    """
    Distance euclidienne entre centroïdes.
    Les centroïdes sont au format (row, col).
    """
    ca = np.asarray(obj_a["centroid"], dtype=float)
    cb = np.asarray(obj_b["centroid"], dtype=float)
    return float(np.linalg.norm(ca - cb))


def mask_iou(mask_a, mask_b):
    """
    Intersection over Union entre deux masques binaires.
    """
    mask_a = np.asarray(mask_a).astype(bool)
    mask_b = np.asarray(mask_b).astype(bool)

    inter = np.logical_and(mask_a, mask_b).sum()
    union = np.logical_or(mask_a, mask_b).sum()

    if union == 0:
        return 0.0

    return float(inter / union)


def transfer_labels_from_position_reference(
    object_db,
    image_db,
    mixture_key="alm1pea1",
    reference_key="pea1_pos1",
    reference_class="peanut",
    other_class="almond",
    min_iou=0.05,
    max_centroid_distance=25,
):
    """
    Transfert automatique des labels depuis une image de référence de position.

    Principe :
    - les objets de reference_key sont des arachides connues ;
    - on les matche aux objets de mixture_key ;
    - les objets matchés deviennent peanut ;
    - les autres objets du mélange deviennent almond.

    Matching :
    score = IoU + petit bonus de proximité centroïde.
    """

    mixture_key = clean_image_key(mixture_key)
    reference_key = clean_image_key(reference_key)

    if mixture_key not in image_db:
        raise KeyError(f"{mixture_key!r} not found in image_db.")

    if reference_key not in image_db:
        raise KeyError(f"{reference_key!r} not found in image_db.")

    mixture_img = image_db[mixture_key]
    reference_img = image_db[reference_key]

    image_shape = mixture_img["labels"].shape

    if reference_img["labels"].shape != image_shape:
        raise ValueError(
            f"Images do not have the same shape: "
            f"{mixture_key}={mixture_img['labels'].shape}, "
            f"{reference_key}={reference_img['labels'].shape}"
        )

    mix_objects = get_objects_from_image(object_db, mixture_key)
    ref_objects_all = get_objects_from_image(object_db, reference_key)

    # On garde uniquement les objets de référence du type voulu.
    # Normalement pea1_pos1 -> peanut.
    ref_objects = {
        oid: obj
        for oid, obj in ref_objects_all.items()
        if obj.get("object_nut_type") == reference_class
        or obj.get("image_nut_type") == reference_class
    }

    if len(ref_objects) == 0:
        raise ValueError(
            f"No reference object of class {reference_class!r} found in {reference_key!r}."
        )

    ref_ids = list(ref_objects.keys())
    mix_ids = list(mix_objects.keys())

    iou_matrix = np.zeros((len(ref_ids), len(mix_ids)), dtype=float)
    dist_matrix = np.zeros((len(ref_ids), len(mix_ids)), dtype=float)

    ref_masks = {
        oid: get_object_global_mask(obj, image_shape)
        for oid, obj in ref_objects.items()
    }

    mix_masks = {
        oid: get_object_global_mask(obj, image_shape)
        for oid, obj in mix_objects.items()
    }

    for i, ref_id in enumerate(ref_ids):
        for j, mix_id in enumerate(mix_ids):
            iou_matrix[i, j] = mask_iou(ref_masks[ref_id], mix_masks[mix_id])
            dist_matrix[i, j] = centroid_distance(ref_objects[ref_id], mix_objects[mix_id])

    # Score combinant recouvrement spatial et proximité.
    # Le terme de distance sert surtout si les masques ne se superposent pas parfaitement.
    distance_bonus = np.clip(
        1 - dist_matrix / max_centroid_distance,
        0,
        1,
    )

    score_matrix = iou_matrix + 0.25 * distance_bonus

    # Hungarian matching : association 1 référence -> 1 objet du mélange.
    ref_ind, mix_ind = linear_sum_assignment(-score_matrix)

    matches = []

    for i, j in zip(ref_ind, mix_ind):
        ref_id = ref_ids[i]
        mix_id = mix_ids[j]

        iou = iou_matrix[i, j]
        dist = dist_matrix[i, j]
        score = score_matrix[i, j]

        accepted = (iou >= min_iou) or (dist <= max_centroid_distance)

        matches.append({
            "reference_object_id": ref_id,
            "mixture_object_id": mix_id,
            "reference_class": reference_class,
            "iou": iou,
            "centroid_distance_px": dist,
            "score": score,
            "accepted": accepted,
            "reference_centroid": ref_objects[ref_id]["centroid"],
            "mixture_centroid": mix_objects[mix_id]["centroid"],
            "reference_area": ref_objects[ref_id]["area_pixels"],
            "mixture_area": mix_objects[mix_id]["area_pixels"],
        })

    matches_df = pd.DataFrame(matches)

    accepted_matches = matches_df[matches_df["accepted"]].copy()

    # Construction du label final des objets du mélange.
    transferred_labels = {
        oid: other_class
        for oid in mix_ids
    }

    for mix_id in accepted_matches["mixture_object_id"]:
        transferred_labels[mix_id] = reference_class

    labels_df = pd.DataFrame([
        {
            "object_id": oid,
            "source_image": mixture_key,
            "transferred_label": transferred_labels[oid],
            "label_id": mix_objects[oid]["label_id"],
            "centroid_row": mix_objects[oid]["centroid"][0],
            "centroid_col": mix_objects[oid]["centroid"][1],
            "area_pixels": mix_objects[oid]["area_pixels"],
            "matched_from_reference": oid in set(accepted_matches["mixture_object_id"]),
        }
        for oid in mix_ids
    ])

    labels_df = labels_df.sort_values(
        ["transferred_label", "object_id"]
    ).reset_index(drop=True)

    return labels_df, matches_df
```

```python
import plotly.graph_objects as go
from pathlib import Path

PRESENTATION_DIR = PROJECT_ROOT / "results" / "presentation_AG"
PRESENTATION_DIR.mkdir(parents=True, exist_ok=True)


def build_class_overlay_from_labels_df(image_db, object_db, image_key, labels_df):
    image_key = clean_image_key(image_key)
    labels_img = image_db[image_key]["labels"]

    overlay = np.zeros_like(labels_img, dtype=int)

    class_to_value = {
        "almond": 1,
        "peanut": 2,
        "unknown": 3,
    }

    for _, row in labels_df.iterrows():
        object_id = row["object_id"]
        nut_class = row["transferred_label"]

        obj = object_db[object_id]
        label_id = int(obj["label_id"])

        overlay[labels_img == label_id] = class_to_value.get(nut_class, 3)

    return overlay


def plot_mixture_transferred_labels(
    image_db,
    object_db,
    image_key,
    labels_df,
    title=None,
    alpha=0.58,
    width=1050,
    height=750,
):
    image_key = clean_image_key(image_key)

    img = image_db[image_key]
    bg = img["image_ref"]

    overlay = build_class_overlay_from_labels_df(
        image_db=image_db,
        object_db=object_db,
        image_key=image_key,
        labels_df=labels_df,
    )

    fig = go.Figure()

    # Fond grayscale
    fig.add_trace(
        go.Heatmap(
            z=bg,
            colorscale="Gray",
            showscale=False,
            hovertemplate="row=%{y}<br>col=%{x}<br>signal=%{z}<extra></extra>",
        )
    )

    # Amandes
    fig.add_trace(
        go.Heatmap(
            z=np.where(overlay == 1, 1, np.nan),
            colorscale=[[0, "#1f77b4"], [1, "#1f77b4"]],
            showscale=False,
            opacity=alpha,
            hovertemplate="Amande<br>row=%{y}<br>col=%{x}<extra></extra>",
        )
    )

    # Arachides
    fig.add_trace(
        go.Heatmap(
            z=np.where(overlay == 2, 1, np.nan),
            colorscale=[[0, "#e74c3c"], [1, "#e74c3c"]],
            showscale=False,
            opacity=alpha,
            hovertemplate="Arachide<br>row=%{y}<br>col=%{x}<extra></extra>",
        )
    )

    # Faux points uniquement pour la légende
    fig.add_trace(
        go.Scatter(
            x=[None],
            y=[None],
            mode="markers",
            marker=dict(size=18, color="#1f77b4"),
            name="Almond",
        )
    )

    fig.add_trace(
        go.Scatter(
            x=[None],
            y=[None],
            mode="markers",
            marker=dict(size=18, color="#e74c3c"),
            name="Peanut",
        )
    )

    fig.update_layout(
        title=title or f"Annotation automatique par image de position — {image_key}",
        template="plotly_white",
        width=width,
        height=height,
        font=dict(size=22),
        margin=dict(l=40, r=40, t=80, b=40),
        legend=dict(
            orientation="h",
            x=0.5,
            xanchor="center",
            y=-0.08,
            yanchor="top",
        ),
    )

    fig.update_xaxes(visible=False)
    fig.update_yaxes(visible=False, autorange="reversed", scaleanchor="x")

    return fig


fig = plot_mixture_transferred_labels(
    image_db=image_db,
    object_db=object_db,
    image_key="alm1pea1",
    labels_df=labels_df,
    title="Mixture image alm1pea1 : identified almonds and peanuts",
)

fig.show()

fig.write_html(PRESENTATION_DIR / "slide2_alm1pea1_transferred_labels.html")

try:
    fig.write_image(PRESENTATION_DIR / "slide2_alm1pea1_transferred_labels.png", scale=2)
except Exception as exc:
    print("PNG export failed. Install kaleido with: pip install -U kaleido")
    print(repr(exc))
```

```python
from pathlib import Path
import numpy as np
import pandas as pd
import plotly.graph_objects as go

from sklearn.decomposition import PCA
from scipy.signal import savgol_filter
from scipy.optimize import linear_sum_assignment

# Dossier export figures
PRESENTATION_DIR = PROJECT_ROOT / "results" / "presentation_AG"
PRESENTATION_DIR.mkdir(parents=True, exist_ok=True)

# Image à analyser
MIXTURE_KEY = "alm1pea1"
REFERENCE_KEY = "pea1_pos1"

# Classes
CONFORM_CLASS = "almond"   # modèle SIMCA appris sur l'amande = classe conforme
TARGET_CLASS = "peanut"    # arachide = anomalie / rejet SIMCA

# Chemin vers les candidats sauvegardés
# Adapte si ton fichier est ailleurs.
CANDIDATE_PATH = PROJECT_ROOT / "results" / "simca_selection"/"selected_simca_candidates.parquet"

if not CANDIDATE_PATH.exists():
    CANDIDATE_PATH = Path("selected_simca_candidates.parquet")

print("Candidate path:", CANDIDATE_PATH)
```

```python
def get_first_available(row, names, default=None):
    for name in names:
        if name in row and pd.notna(row[name]):
            return row[name]
    return default


preprocessing_name = get_first_available(
    candidate,
    ["preprocessing", "preprocessing_method", "preprocess", "preproc"],
    default="absorbance_sg_d1",
)

n_components = int(get_first_available(
    candidate,
    ["n_components", "n_comp", "n_pc", "n_pcs"],
    default=7,
))

rule_name = str(get_first_available(
    candidate,
    ["rule", "rule_variant", "selected_rule_name", "rule_for_refit"],
    default="data_driven_emp_cv",
))

print("Used preprocessing:", preprocessing_name)
print("Used n_components:", n_components)
print("Used rule:", rule_name)
```

```python
def clean_key(key):
    """
    Harmonise les noms : alm1_pea1 -> alm1pea1, suffixe _sb retiré.
    """
    key = str(key).strip().lower()
    key = key.replace("_pea", "pea").replace("_alm", "alm").replace("_wal", "wal")
    key = key.replace("_", "")
    if key.endswith("sb"):
        key = key[:-2]
    return key


def resolve_image_key(image_db, requested_key):
    """
    Retrouve une clé image_db même si l'utilisateur écrit alm1_pea1 au lieu de alm1pea1.
    """
    requested_clean = clean_key(requested_key)

    for key in image_db.keys():
        if clean_key(key) == requested_clean:
            return key

    raise KeyError(
        f"Image {requested_key!r} not found. Available close keys: "
        f"{[k for k in image_db.keys() if requested_clean[:3] in clean_key(k)][:10]}"
    )


def get_objects_from_image(object_db, image_key):
    image_key_clean = clean_key(image_key)

    out = {
        oid: obj
        for oid, obj in object_db.items()
        if clean_key(obj.get("source_clean_key", obj.get("source_image", ""))) == image_key_clean
    }

    if len(out) == 0:
        raise ValueError(f"No object found for image {image_key!r}")

    return out


def safe_log_absorbance(X, eps=1e-8):
    X = np.asarray(X, dtype=float)
    X = np.clip(X, eps, None)
    return -np.log10(X)


def snv(X, eps=1e-12):
    X = np.asarray(X, dtype=float)
    mu = np.nanmean(X, axis=1, keepdims=True)
    sd = np.nanstd(X, axis=1, keepdims=True)
    return (X - mu) / np.maximum(sd, eps)


def preprocess_spectra_for_candidate(X, preprocessing):
    """
    Prétraitements courants du projet.
    Le même traitement est appliqué au train et aux pixels projetés.
    """
    Xp = np.asarray(X, dtype=float)
    name = str(preprocessing).lower()

    if "absorbance" in name:
        Xp = safe_log_absorbance(Xp)

    if "snv" in name:
        Xp = snv(Xp)

    # Savitzky-Golay
    if "sg" in name or "savgol" in name:
        # fenêtre impaire et compatible avec le nombre de bandes
        window = 15
        if Xp.shape[1] < window:
            window = Xp.shape[1] if Xp.shape[1] % 2 == 1 else Xp.shape[1] - 1
        window = max(window, 5)

        deriv = 0
        if "d1" in name or "deriv1" in name or "first" in name:
            deriv = 1
        elif "d2" in name or "deriv2" in name or "second" in name:
            deriv = 2

        Xp = savgol_filter(
            Xp,
            window_length=window,
            polyorder=2,
            deriv=deriv,
            axis=1,
            mode="interp",
        )

    return Xp


def fit_simca_one_class(X_train_raw, preprocessing, n_components, h_quantile=0.95, q_quantile=0.95):
    """
    SIMCA simple :
    - prétraitement,
    - centrage,
    - PCA,
    - H = Hotelling-like score distance,
    - Q = résidu SPE,
    - limites empiriques sur le train.
    """
    X_train = preprocess_spectra_for_candidate(X_train_raw, preprocessing)

    finite_mask = np.all(np.isfinite(X_train), axis=1)
    X_train = X_train[finite_mask]

    x_mean = np.nanmean(X_train, axis=0)
    Xc = X_train - x_mean

    pca = PCA(n_components=n_components, random_state=42)
    T = pca.fit_transform(Xc)

    X_hat = pca.inverse_transform(T)
    E = Xc - X_hat

    Q_train = np.sum(E ** 2, axis=1)
    H_train = np.sum((T ** 2) / pca.explained_variance_, axis=1)

    H_limit = np.nanquantile(H_train, h_quantile)
    Q_limit = np.nanquantile(Q_train, q_quantile)

    model = {
        "preprocessing": preprocessing,
        "n_components": n_components,
        "x_mean": x_mean,
        "pca": pca,
        "H_limit": H_limit,
        "Q_limit": Q_limit,
        "H_train": H_train,
        "Q_train": Q_train,
    }

    return model


def predict_simca_one_class(model, X_raw, rule="simple"):
    """
    Retourne accepted=True si le pixel est accepté par le modèle amande.
    Donc predicted_peanut = ~accepted.
    """
    X = preprocess_spectra_for_candidate(X_raw, model["preprocessing"])
    Xc = X - model["x_mean"]

    pca = model["pca"]
    T = pca.transform(Xc)

    X_hat = pca.inverse_transform(T)
    E = Xc - X_hat

    Q = np.sum(E ** 2, axis=1)
    H = np.sum((T ** 2) / pca.explained_variance_, axis=1)

    Hn = H / model["H_limit"]
    Qn = Q / model["Q_limit"]

    rule = str(rule).lower()

    if "alternative" in rule or "sum" in rule:
        rule_statistic = Hn + Qn
        rule_limit = 2.0
        accepted = rule_statistic <= rule_limit

    elif "data" in rule or "emp" in rule or "cv" in rule:
        # Version simple et robuste pour visualisation :
        # max(H/Hlim, Q/Qlim) <= 1
        rule_statistic = np.maximum(Hn, Qn)
        rule_limit = 1.0
        accepted = rule_statistic <= rule_limit

    else:
        # simple SIMCA
        rule_statistic = np.maximum(Hn, Qn)
        rule_limit = 1.0
        accepted = (Hn <= 1.0) & (Qn <= 1.0)

    return {
        "accepted": accepted,
        "predicted_peanut": ~accepted,
        "H": H,
        "Q": Q,
        "H_norm_limit": Hn,
        "Q_norm_limit": Qn,
        "rule_statistic": rule_statistic,
        "rule_limit": rule_limit,
    }
```

```python
def collect_training_pixels(
    object_db,
    conform_class="almond",
    split_values=None,
    max_pixels=80000,
    random_state=42,
):
    """
    Récupère les pixels de calibration pour la classe conforme.
    Par défaut : objets purs d'amande.
    """
    X_list = []

    for oid, obj in object_db.items():
        is_class = obj.get("object_nut_type") == conform_class
        is_pure = bool(obj.get("is_pure", False))

        if not (is_class and is_pure):
            continue

        if split_values is not None:
            if obj.get("split") not in split_values:
                continue

        if "spectra" not in obj:
            continue

        X_list.append(np.asarray(obj["spectra"], dtype=float))

    if len(X_list) == 0:
        raise RuntimeError(f"No pure pixels found for conform_class={conform_class!r}")

    X = np.vstack(X_list)

    if len(X) > max_pixels:
        rng = np.random.default_rng(random_state)
        idx = rng.choice(len(X), size=max_pixels, replace=False)
        X = X[idx]

    return X


X_train_raw = collect_training_pixels(
    object_db=object_db,
    conform_class=CONFORM_CLASS,
    split_values=None,
    max_pixels=80000,
    random_state=42,
)

print("X_train_raw:", X_train_raw.shape)

simca_model = fit_simca_one_class(
    X_train_raw=X_train_raw,
    preprocessing=preprocessing_name,
    n_components=n_components,
    h_quantile=0.95,
    q_quantile=0.95,
)

print("SIMCA fitted")
print("H_limit:", simca_model["H_limit"])
print("Q_limit:", simca_model["Q_limit"])
```

```python
def get_global_mask(obj, image_shape):
    if "mask_global" in obj and obj["mask_global"] is not None:
        return np.asarray(obj["mask_global"]).astype(bool)

    min_row, min_col, max_row, max_col = map(int, obj["bbox"])
    mask_crop = np.asarray(obj["mask"]).astype(bool)

    out = np.zeros(image_shape, dtype=bool)
    out[min_row:max_row, min_col:max_col] = mask_crop
    return out


def mask_iou(a, b):
    a = np.asarray(a).astype(bool)
    b = np.asarray(b).astype(bool)
    inter = np.logical_and(a, b).sum()
    union = np.logical_or(a, b).sum()
    return 0.0 if union == 0 else float(inter / union)


def centroid_distance(obj_a, obj_b):
    ca = np.asarray(obj_a["centroid"], dtype=float)
    cb = np.asarray(obj_b["centroid"], dtype=float)
    return float(np.linalg.norm(ca - cb))


def transfer_truth_from_position_reference(
    object_db,
    image_db,
    mixture_key,
    reference_key,
    reference_class="peanut",
    other_class="almond",
    min_iou=0.03,
    max_centroid_distance=30,
):
    mixture_key = resolve_image_key(image_db, mixture_key)
    reference_key = resolve_image_key(image_db, reference_key)

    mix_img = image_db[mixture_key]
    ref_img = image_db[reference_key]

    if mix_img["labels"].shape != ref_img["labels"].shape:
        raise ValueError(
            f"Shape mismatch: {mixture_key}={mix_img['labels'].shape}, "
            f"{reference_key}={ref_img['labels'].shape}"
        )

    image_shape = mix_img["labels"].shape

    mix_objects = get_objects_from_image(object_db, mixture_key)
    ref_objects_all = get_objects_from_image(object_db, reference_key)

    ref_objects = {
        oid: obj
        for oid, obj in ref_objects_all.items()
        if obj.get("object_nut_type") == reference_class
        or obj.get("image_nut_type") == reference_class
    }

    if len(ref_objects) == 0:
        raise RuntimeError(f"No {reference_class} object found in {reference_key}")

    mix_ids = list(mix_objects.keys())
    ref_ids = list(ref_objects.keys())

    mix_masks = {
        oid: get_global_mask(obj, image_shape)
        for oid, obj in mix_objects.items()
    }
    ref_masks = {
        oid: get_global_mask(obj, image_shape)
        for oid, obj in ref_objects.items()
    }

    iou_matrix = np.zeros((len(ref_ids), len(mix_ids)))
    dist_matrix = np.zeros((len(ref_ids), len(mix_ids)))

    for i, rid in enumerate(ref_ids):
        for j, mid in enumerate(mix_ids):
            iou_matrix[i, j] = mask_iou(ref_masks[rid], mix_masks[mid])
            dist_matrix[i, j] = centroid_distance(ref_objects[rid], mix_objects[mid])

    distance_bonus = np.clip(1 - dist_matrix / max_centroid_distance, 0, 1)
    score_matrix = iou_matrix + 0.25 * distance_bonus

    ref_ind, mix_ind = linear_sum_assignment(-score_matrix)

    rows = []
    matched_mix_ids = set()

    for i, j in zip(ref_ind, mix_ind):
        rid = ref_ids[i]
        mid = mix_ids[j]

        iou = iou_matrix[i, j]
        dist = dist_matrix[i, j]
        score = score_matrix[i, j]

        accepted = (iou >= min_iou) or (dist <= max_centroid_distance)

        rows.append({
            "reference_object_id": rid,
            "mixture_object_id": mid,
            "iou": iou,
            "centroid_distance_px": dist,
            "score": score,
            "accepted": accepted,
        })

        if accepted:
            matched_mix_ids.add(mid)

    matches_df = pd.DataFrame(rows)

    truth_rows = []
    for mid in mix_ids:
        obj = mix_objects[mid]
        truth = reference_class if mid in matched_mix_ids else other_class

        truth_rows.append({
            "object_id": mid,
            "true_label": truth,
            "label_id": int(obj["label_id"]),
            "centroid": obj["centroid"],
            "area_pixels": obj["area_pixels"],
            "source_image": mixture_key,
        })

    truth_objects_df = pd.DataFrame(truth_rows)

    return truth_objects_df, matches_df


truth_objects_df, matches_df = transfer_truth_from_position_reference(
    object_db=object_db,
    image_db=image_db,
    mixture_key=MIXTURE_KEY,
    reference_key=REFERENCE_KEY,
    reference_class=TARGET_CLASS,
    other_class=CONFORM_CLASS,
    min_iou=0.03,
    max_centroid_distance=30,
)

display(matches_df.sort_values("score", ascending=False))
display(truth_objects_df)
display(truth_objects_df["true_label"].value_counts())
```

```python
def make_case_maps(pixel_df, image_shape):
    maps = {
        "TP": np.zeros(image_shape, dtype=int),
        "TN": np.zeros(image_shape, dtype=int),
        "FP": np.zeros(image_shape, dtype=int),
        "FN": np.zeros(image_shape, dtype=int),
    }

    for case_name in maps:
        sub = pixel_df[pixel_df["case"].eq(case_name)]
        maps[case_name][sub["row"].to_numpy(), sub["col"].to_numpy()] = 1

    return maps


case_maps = make_case_maps(pixel_df, labels_img.shape)

print({k: int(v.sum()) for k, v in case_maps.items()})
```

```python
def plot_fp_fn_overlay(
    image_db,
    image_key,
    pixel_df,
    case_maps,
    title=None,
    show_tp=True,
    alpha_tp=0.25,
    alpha_errors=0.90,
    width=1200,
    height=850,
):
    image_key = resolve_image_key(image_db, image_key)
    img = image_db[image_key]

    bg = np.asarray(img["image_ref"], dtype=float)

    n_fp = int(case_maps["FP"].sum())
    n_fn = int(case_maps["FN"].sum())
    n_tp = int(case_maps["TP"].sum())
    n_tn = int(case_maps["TN"].sum())

    fig = go.Figure()

    # Fond image
    fig.add_trace(
        go.Heatmap(
            z=bg,
            colorscale="Gray",
            showscale=False,
            hovertemplate="row=%{y}<br>col=%{x}<br>signal=%{z}<extra></extra>",
            name="Image",
        )
    )

    # TP en rouge transparent : arachides détectées correctement
    if show_tp:
        fig.add_trace(
            go.Heatmap(
                z=np.where(case_maps["TP"] == 1, 1, np.nan),
                colorscale=[[0, "rgba(231,76,60,1)"], [1, "rgba(231,76,60,1)"]],
                opacity=alpha_tp,
                showscale=False,
                hovertemplate="TP : arachide détectée<br>row=%{y}<br>col=%{x}<extra></extra>",
                name="TP",
            )
        )

    # FP : amande prédite arachide
    fig.add_trace(
        go.Heatmap(
            z=np.where(case_maps["FP"] == 1, 1, np.nan),
            colorscale=[[0, "rgba(155,89,182,1)"], [1, "rgba(155,89,182,1)"]],
            opacity=alpha_errors,
            showscale=False,
            hovertemplate="FP : amande prédite arachide<br>row=%{y}<br>col=%{x}<extra></extra>",
            name="FP",
        )
    )

    # FN : arachide prédite amande
    fig.add_trace(
        go.Heatmap(
            z=np.where(case_maps["FN"] == 1, 1, np.nan),
            colorscale=[[0, "rgba(241,196,15,1)"], [1, "rgba(241,196,15,1)"]],
            opacity=alpha_errors,
            showscale=False,
            hovertemplate="FN : arachide ratée<br>row=%{y}<br>col=%{x}<extra></extra>",
            name="FN",
        )
    )

    # Faux points pour légende lisible
    legend_items = [
        ("TP arachide détectée", "rgba(231,76,60,1)") if show_tp else None,
        (f"FP fausse alerte ({n_fp})", "rgba(155,89,182,1)"),
        (f"FN arachide ratée ({n_fn})", "rgba(241,196,15,1)"),
    ]

    for item in legend_items:
        if item is None:
            continue
        label, color = item
        fig.add_trace(
            go.Scatter(
                x=[None],
                y=[None],
                mode="markers",
                marker=dict(size=18, color=color),
                name=label,
            )
        )

    fig.update_layout(
        title=title or (
            f"Erreurs pixel SIMCA — {image_key}<br>"
            f"<sup>TN={n_tn}, TP={n_tp}, FP={n_fp}, FN={n_fn}</sup>"
        ),
        template="plotly_white",
        width=width,
        height=height,
        font=dict(size=22),
        margin=dict(l=40, r=40, t=100, b=80),
        legend=dict(
            orientation="h",
            x=0.5,
            xanchor="center",
            y=-0.08,
            yanchor="top",
        ),
    )

    fig.update_xaxes(visible=False)
    fig.update_yaxes(visible=False, autorange="reversed", scaleanchor="x")

    return fig


fig = plot_fp_fn_overlay(
    image_db=image_db,
    image_key=MIXTURE_KEY,
    pixel_df=pixel_df,
    case_maps=case_maps,
    title=(
        "Carte des erreurs pixel — SIMCA amande sur image mixte alm1pea1<br>"
        f"<sup>Prétraitement={preprocessing_name}, n_components={n_components}, règle={rule_name}</sup>"
    ),
    show_tp=True,
)

fig.show()

# html_path = PRESENTATION_DIR / "alm1pea1_simca_fp_fn_overlay.html"
# png_path = PRESENTATION_DIR / "alm1pea1_simca_fp_fn_overlay.png"

# fig.write_html(html_path)

# try:
#     fig.write_image(png_path, scale=2)
#     print("PNG saved:", png_path)
# except Exception as exc:
#     print("PNG export failed. Install kaleido with: pip install -U kaleido")
#     print(repr(exc))

# print("HTML saved:", html_path)
```
