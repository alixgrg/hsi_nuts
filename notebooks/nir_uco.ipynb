### Savitzky-Golay Derivatives

```python
delta = np.mean(np.diff(wavelengths))
X_d1 = savgol_derivative(
    X_mean_objects,
    window_length=9,
    polyorder=2,
    deriv=1,
    delta=delta,
)
X_d1_c, mu_d1 = center_X(X_d1)

pca_mean_sg = pca_from_cov(
    X_d1_c,
    n_components=6,
)

T_mean_sg = pca_mean_sg["scores"]
P_mean_sg = pca_mean_sg["loadings"]

print("X:", X_d1.shape)
print("X centered:", X_d1_c.shape)
print("Scores:", T_mean_sg.shape)
print("Loadings:", P_mean_sg.shape)
print("Explained variance:", pca_mean_sg["explained_variance_ratio"][:6])
```

```python
plot_pca_scores_2d(
    pca_res=pca_mean_sg,
    labels=y_objects,
    object_ids=object_ids,
    source_images=source_images,
    batches=batches,
    areas=areas,
    pcx=1,
    pcy=2,
    color_by="label",
)

plot_pca_loadings(pca_res=pca_mean_sg, wavelengths=wavelengths, components=(1,2,3,4))

plot_pca_biplot_2d(
    pca_res=pca_mean_sg,
    labels=y_objects,
    object_ids=object_ids,
    wavelengths=wavelengths,
    n_loadings=4)

plot_pca_q_vs_t2(X_d1_c, pca_res=pca_mean_sg, labels=y_objects, object_ids=object_ids, source_images=source_images, n_components=4)
```
