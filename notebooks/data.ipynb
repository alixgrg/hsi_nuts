# HSI DATA ANALYZING

**1- NIR camera UCO (889-170nm)**
- "NIR_uco_sb.mat": nuts data after black/white correction and without background. There are 2 types of nut (almond and peanut). Per nut type, 3 different batches were bought. Almond 1, Almond 2, Almond 3, Peanut 1, Penaut 2, Peanut 3. Noisy areas of the spectrum have NOT been removed yet. First 6 wavelengths out of the 69 should be removed. Data in reflectance mode. From University of Cordoba.

**2- NIR camera USDA (419-1007nm)**
- "Loadings_nir.xlsx": loadings info of the Project developed with the USDA to discriminate between Almonds, hazelnuts, walnuts and peanuts
- "Mean spectrum per nut piece NIR.xlsx": 2 sheets, one per independent set of nuts. Each sheet contains the mean spectrum per nut piece (100 pieces x 4 nut types). Noisy areas of the spectrum have been already removed. Range considered --> 443-973 nm. Data in absorbance --> Log10(1/reflectance).
- "NIR_sb.mat": nuts data after black/white correction and without background, data in reflectance mode. There are 4 types of nut (almond, hazelnut, walnut and peanut), There are 2 independent sets per nut type. Total, 8 images.

**3- SWIR camera USDA (842-2532nm)**
- "Loadings_swir.xlsx": loadings info of the Project developed with the USDA to discriminate between Almonds, hazelnuts, walnuts and peanuts.
- "Mean spectrum per nut piece SWIR.xlsx": 2 sheets, one per independent set of nuts. Each sheet contains the mean spectrum per nut piece (100 pieces x 4 nut types). Noisy areas of the spectrum have been already removed. Range considered --> 930-2384 nm. Data in absorbance --> Log10(1/refletance)
- "SWIR_sb.mat": nuts data after black/white correction and without background. There are 4 types of nut (almond, hazelnut, walnut and peanut), There are 2 independent sets per nut type. Total, 8 images. Data in reflectance.

- correction noir/ blanc = enlever les effets instrumentaux
$$
R(x,y,\lambda) = \frac{I_{raw}(x,y,\lambda)-I_{dark}(\lambda)}{I_{white}(\lambda)-I_{dark}(\lambda)}
$$
- "reflectance mode": instrument éclaire échantillon puis mesure la lumière renvoyée par sa surface ou proche surface
$$
R(\lambda) = \frac{\text{lumière réfléchie par l'échantillon à }\lambda}{\text{lumière réfléchie par une référence blanche à }\lambda}
$$
donc cubes hyperspectraux contiennent des valeurs de réflectance et pour chaque pixel, on a une corube de rélectance en fonction de la longueur d'onde.
- SWIR = "Short-Wave Infrared"

```python
def save_pickle(obj, path):
    """
    Save Python object as pickle.
    """
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)
    with open(path, "wb") as f:
        pickle.dump(obj, f)
    print(f"[OK] Saved: {path}")
```

```python
full_minimal_object_db = {}
full_minimal_object_db.update(object_db)
full_minimal_object_db.update(object_proj_db)
save_pickle(full_minimal_object_db, PROJECT_ROOT / "HSI Data" / "processed" / "nir_uco_full_minimal_object_db_reflectance.pkl")
```
