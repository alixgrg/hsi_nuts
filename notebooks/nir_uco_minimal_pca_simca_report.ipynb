## From object database to modelling matrix

For object-level modelling, we build:

$$
X \in \mathbb{R}^{N \times B}
$$

where:
- N is the number of objects;
- B is the number of spectral bands;
- each row corresponds to one object mean spectrum

We also keep:
- y: the object labels;
- meta: metadata aligned with the rows of X

We use src/redim_matrix/ object_db_to_matrix function:
- takes the object db as input
- returns X, y, meta
- different observaiton levels available (object, pixel, balanced_pixel)
- filters to elect a specific subset of objects (e.g. batch 1 and 2 are for training, 3 and 4 for external projection, only pure objects are used, mixture objects are not used for supervised training)

(different observation levels were implemented as tests, best results obtained using level='object')

## Pre-processing

Input matrices:
- X_train_raw: object spectra from batches 1 and 2;
- y_train: labels associated with the training objects;
- meta_train: metadata aligned with X_train_raw;
- X_projection_raw: object spectra from batches 3 and 4;
- y_projection: labels associated with the projection objects;
- meta_projection: metadata aligned with X_projection_raw.

The pre-processing methods are compared on the same train/projection split

We use src/preprocessing/ SpectralPreprocessor class

| Name | Steps | Purpose |
|---|---|---|
| `raw` | no transformation | Baseline reference |
| `absorbance` | reflectance to absorbance | Convert reflectance-like signal into absorbance-like signal |
| `snv` | Standard Normal Variate | Correct spectrum-wise offset and scaling effects |
| `vector_norm` | L2 normalization | Normalize each spectrum by its vector norm |
| `msc` | Multiplicative Scatter Correction | Correct additive and multiplicative scatter effects |
| `absorbance_snv` | absorbance + SNV | Combine absorbance transform and scatter normalization |
| `absorbance_sg_d1` | absorbance + first derivative | Highlight local spectral variations |
| `absorbance_snv_sg_d1` | absorbance + SNV + first derivative | Combine scatter correction and derivative information |


For each pre-processing method, the workflow is:

- preprocessor.fit(X_train_raw, wavelengths=wavelengths)
- X_train_preprocessed = preprocessor.transform(X_train_raw)
- X_projection_preprocessed = preprocessor.transform(X_projection_raw)


For each pre-processing method, the following objects are available:
- preprocessed_data[name]["X_train"]
- preprocessed_data[name]["X_projection"]
- preprocessed_data[name]["y_train"]
- preprocessed_data[name]["y_projection"]
- preprocessed_data[name]["meta_train"]
- preprocessed_data[name]["meta_projection"]
- preprocessed_data[name]["preprocessor"]

```python
preprocessing_configs = {
    "raw": ("raw",),
    "absorbance": ("absorbance",),
    "snv": ("snv",),
    "vector_norm": ("vector_norm",),
    "msc": ("msc",),
    "absorbance_snv": ("absorbance", "snv"),
    "absorbance_sg_d1": ("absorbance", "sg_d1"),
    "absorbance_snv_sg_d1": ("absorbance", "snv", "sg_d1"),
}
```

```python
# Apply all preprocessing configurations to training and projection data
preprocessed_data = {}

for name, steps in preprocessing_configs.items():
    print(f"Applying preprocessing: {name} | steps={steps}")
    
    preprocessor = SpectralPreprocessor(
        steps=steps,
        sg_window_length=9,
        sg_polyorder=2,
    )
    
    X_train_pre = preprocessor.fit_transform(
        X_train_raw,
        wavelengths=wavelengths,
    )
    
    X_projection_pre = preprocessor.transform(
        X_projection_raw,
    )
    
    preprocessed_data[name] = {
        "steps": steps,
        "preprocessor": preprocessor,
        "X_train": X_train_pre,
        "X_projection": X_projection_pre,
        "y_train": y_train,
        "y_projection": y_projection,
        "meta_train": meta_train,
        "meta_projection": meta_projection,
    }

print("\nDone.")
```

### Quantitative comparison

```python
def preprocessing_effect_metrics(X):
    """
    Compute simple descriptive metrics for a preprocessed spectral matrix.
    """
    X = np.asarray(X, dtype=float)
    
    spectrum_means = np.mean(X, axis=1)
    spectrum_stds = np.std(X, axis=1, ddof=1)
    
    wavelength_stds = np.std(X, axis=0, ddof=1)
    wavelength_ranges = np.nanmax(X, axis=0) - np.nanmin(X, axis=0)
    
    return {
        "mean_abs_value": float(np.mean(np.abs(X))),
        "mean_spectrum_mean": float(np.mean(spectrum_means)),
        "mean_spectrum_std": float(np.mean(spectrum_stds)),
        "mean_wavelength_std": float(np.mean(wavelength_stds)),
        "mean_wavelength_range": float(np.mean(wavelength_ranges)),
        "global_min": float(np.nanmin(X)),
        "global_max": float(np.nanmax(X)),
    }


rows = []

for name, item in preprocessed_data.items():
    train_metrics = preprocessing_effect_metrics(item["X_train"])
    projection_metrics = preprocessing_effect_metrics(item["X_projection"])
    
    rows.append({
        "preprocessing": name,
        "subset": "train",
        **train_metrics,
    })
    
    rows.append({
        "preprocessing": name,
        "subset": "projection",
        **projection_metrics,
    })

preprocessing_metrics_df = pd.DataFrame(rows)
display(preprocessing_metrics_df)
```

## PCA

PCA is fitted only on the training set:
- almond and peanut pure objects;
- batches 1 and 2 only.

Then, batches 3 and 4 are projected into the PCA space learned from batches 1 and 2.

This allows us to answer three questions:

1. Are almond and peanut objects separated in PCA score space?
2. Do batches 3 and 4 project consistently into the space learned from batches 1 and 2?
3. Are batches 3 and 4 well reconstructed by the PCA model trained on batches 1 and 2?


We use:
- src/pca/ PCAModel class
- src/pca_compison/ class_separation_scores function

PCA reminder:

PCA first centers the matrix:
$$
X_c = X - \mu
$$

Then it computes the covariance matrix:
$$
S = \frac{X_c^\top X_c}{N - 1}
$$

The principal components are obtained from the eigendecomposition:
$$
S p_a = \lambda_a p_a
$$

where:

- p_a is the loading vector of component a;
- lambda_a is the variance explained by component a.

The scores are obtained by projection:
$$
T = X_c P
$$

and the reconstruction using the first A components is:
$$
\hat{X}_c = T_A P_A^\top
$$

The residual matrix is:
$$
E = X_c - \hat{X}_c
$$

### PCAModel class

PCA model is fitted on X_train and then used to project X_projection.

Typical usage:
- pca = PCAModel(n_components=10, center=True)
- pca.fit(X_train)
- T_train = pca.transform(X_train)
- T_projection = pca.transform(X_projection)
- X_projection_hat = pca.inverse_transform(T_projection)

The model stores:
- mean_: training mean spectrum;
- loadings_: PCA loading vectors;
- scores_: training scores;
- eigenvalues_: component variances;
- explained_variance_ratio_: explained variance ratio;
- cumulative_explained_variance_ratio_: cumulative explained variance

#### RMSE

To evaluate how well batches 3 and 4 are reconstructed, we compute the RMSE between each pre-processed spectrum and its PCA reconstruction.

For object i:

$$
RMSE_i = \sqrt{\frac{1}{B} \sum_{b=1}^{B} \left(x_{i,b} - \hat{x}_{i,b}\right)^2}
$$

A larger RMSE indicates that the object is less well represented by the PCA model trained on batches 1 and 2

## SIMCA

### SIMCA reminder

For one class c, SIMCA fits a PCA model using only the training spectra of that class.

For a new spectrum x, the model computes:
$$
x_c = x - \mu_c
$$

$$
t = x_c P_c
$$

$$
\hat{x}_c = t P_c^\top
$$

$$
e = x_c - \hat{x}_c
$$

Two distances are then computed.
$$
H = \sum_{a=1}^{A} \frac{t_a^2}{\lambda_a}
$$
where t_a is the score on component a, and lambda_a is the corresponding eigenvalue.


$$
Q = \sum_{b=1}^{B} e_b^2
$$

The object is accepted by the class if the selected SIMCA rule considers its H and Q values compatible with the class model

We use src/simca/ SIMCAClassifier, SimpleSIMCARule, AltSIMCARule, CombinedIndexSIMCARule, DataDrivenSIMCARule classes

SIMCAClassifier typical use:
- simca = SIMCAClassifier(
    class_names=["almond", "peanut"],
    n_components_by_class={"almond": 3, "peanut": 3},
    alpha=0.05,
    rule=SimpleSIMCARule(),
)
- simca.fit(X_train, y_train)
- predictions, closest_classes, simca_results = simca.predict(X_projection)

outputs:
- predictions: final SIMCA decision;
- closest_classes: closest class according to normalized class distance;
- simca_results: detailed distances and acceptance values for each class.

```python
def evaluate_simca_results(results_df, positive_class="peanut", negative_class="almond"):
    """
    Evaluate SIMCA results on a labelled projection set.

    The function treats only exact class predictions as correct.
    Unknown and ambiguous decisions are reported separately.
    """
    df = results_df.copy()
    n = len(df)
    
    if n == 0:
        raise ValueError("results_df is empty.")
    
    pred = df["prediction"].to_numpy()
    true = df["true_label"].to_numpy()
    
    exact_correct = pred == true
    
    unknown_rate = np.mean(pred == "unknown")
    ambiguous_rate = np.mean(pred == "ambiguous")
    exact_accuracy = np.mean(exact_correct)
    
    metrics = {
        "n_objects": n,
        "exact_accuracy": exact_accuracy,
        "unknown_rate": unknown_rate,
        "ambiguous_rate": ambiguous_rate,
    }
    
    for cls in sorted(df["true_label"].unique()):
        cls_mask = true == cls
        other_mask = true != cls
        
        metrics[f"{cls}_n"] = int(cls_mask.sum())
        metrics[f"{cls}_sensitivity"] = (
            np.mean(pred[cls_mask] == cls)
            if cls_mask.sum() > 0 else np.nan
        )
        metrics[f"{cls}_false_positive_rate"] = (
            np.mean(pred[other_mask] == cls)
            if other_mask.sum() > 0 else np.nan
        )
        metrics[f"{cls}_specificity"] = (
            1.0 - metrics[f"{cls}_false_positive_rate"]
            if not np.isnan(metrics[f"{cls}_false_positive_rate"]) else np.nan
        )
    
    # Specific focus on peanut detection
    pos_mask = true == positive_class
    neg_mask = true == negative_class
    
    metrics[f"{positive_class}_detected_as_{positive_class}"] = (
        np.mean(pred[pos_mask] == positive_class)
        if pos_mask.sum() > 0 else np.nan
    )
    metrics[f"{negative_class}_misdetected_as_{positive_class}"] = (
        np.mean(pred[neg_mask] == positive_class)
        if neg_mask.sum() > 0 else np.nan
    )
    
    return metrics


current_simca_metrics = evaluate_simca_results(
    projection_results_df,
    positive_class="peanut",
    negative_class="almond",
)

current_simca_metrics_df = pd.DataFrame([current_simca_metrics])
display(current_simca_metrics_df.T.rename(columns={0: "value"}))
```

### SIMCA rule comparison

Different SIMCA rules lead to different acceptance regions.

We compare four rules:

| Rule | Principle |
|---|---|
| `simple` | accept if both \(H < H_{\text{limit}}\) and \(Q < Q_{\text{limit}}\) |
| `alternative` | accept if \(H/H_{\text{limit}} + Q/Q_{\text{limit}} < 2\) |
| `combined_index` | use a scaled chi-square approximation fitted on training combined distances |
| `data_driven` | use DD-SIMCA style statistic based on \(H_0, Q_0, N_H, N_Q\) |

This comparison is important because a stricter rule may reject more objects, while a more permissive rule may increase false positive or ambiguous decisions

```python
def make_simca_rules():
    return {
        "simple": SimpleSIMCARule(),
        "alternative": AltSIMCARule(threshold=2.0),
        "combined_index": CombinedIndexSIMCARule(),
        "data_driven": DataDrivenSIMCARule(),
    }
```

```python
simca_rule_results = {}
simca_rule_summary_rows = []

for rule_name, rule in make_simca_rules().items():
    clf = SIMCAClassifier(
        class_names=class_names,
        n_components_by_class=simca_n_components_by_class,
        alpha=alpha,
        rule=rule,
    )
    
    clf.fit(X_train, y_train)
    y_pred, closest, detailed_results = clf.predict(X_projection)
    
    res_df = meta_projection_df.copy()
    res_df["true_label"] = y_projection
    res_df["prediction"] = y_pred
    res_df["closest_class"] = closest
    res_df["preprocessing"] = selected_preprocessing
    res_df["rule"] = rule_name
    res_df["simca_case"] = np.where(
        y_pred == "unknown",
        "unknown",
        np.where(
            y_pred == "ambiguous",
            "ambiguous",
            y_pred + "_only",
        )
    )
    
    metrics = evaluate_simca_results(res_df)
    metrics["preprocessing"] = selected_preprocessing
    metrics["rule"] = rule_name
    
    simca_rule_summary_rows.append(metrics)
    simca_rule_results[rule_name] = {
        "classifier": clf,
        "results_df": res_df,
        "detailed_results": detailed_results,
        "metrics": metrics,
    }

simca_rule_summary_df = pd.DataFrame(simca_rule_summary_rows)

display(
    simca_rule_summary_df[
        [
            "preprocessing",
            "rule",
            "n_objects",
            "exact_accuracy",
            "almond_sensitivity",
            "peanut_sensitivity",
            "almond_specificity",
            "peanut_specificity",
            "unknown_rate",
            "ambiguous_rate",
        ]
    ].sort_values("peanut_sensitivity", ascending=False)
)
```

- unknown rate;
- ambiguous rate.

```python
simca_all_results = {}
simca_summary_rows = []

# Use PCA-selected candidates if available, otherwise use all preprocessing candidates
simca_candidate_preprocessings = (
    pca_candidate_preprocessings
    if "pca_candidate_preprocessings" in globals()
    else candidate_preprocessings
)

for prep_name in simca_candidate_preprocessings:
    item = preprocessed_data[prep_name]
    
    Xtr = item["X_train"]
    Xpr = item["X_projection"]
    ytr = item["y_train"]
    ypr = item["y_projection"]
    
    meta_pr = item["meta_projection"]
    if isinstance(meta_pr, dict):
        meta_pr_df = pd.DataFrame(meta_pr)
    else:
        meta_pr_df = meta_pr.copy()
    
    simca_all_results[prep_name] = {}
    
    for rule_name, rule in make_simca_rules().items():
        clf = SIMCAClassifier(
            class_names=class_names,
            n_components_by_class=simca_n_components_by_class,
            alpha=alpha,
            rule=rule,
        )
        
        clf.fit(Xtr, ytr)
        y_pred, closest, detailed_results = clf.predict(Xpr)
        
        res_df = meta_pr_df.copy()
        res_df["true_label"] = ypr
        res_df["prediction"] = y_pred
        res_df["closest_class"] = closest
        res_df["preprocessing"] = prep_name
        res_df["rule"] = rule_name
        res_df["simca_case"] = np.where(
            y_pred == "unknown",
            "unknown",
            np.where(
                y_pred == "ambiguous",
                "ambiguous",
                y_pred + "_only",
            )
        )
        
        metrics = evaluate_simca_results(res_df)
        metrics["preprocessing"] = prep_name
        metrics["rule"] = rule_name
        
        simca_summary_rows.append(metrics)
        
        simca_all_results[prep_name][rule_name] = {
            "classifier": clf,
            "results_df": res_df,
            "detailed_results": detailed_results,
            "metrics": metrics,
        }

simca_summary_df = pd.DataFrame(simca_summary_rows)

display(
    simca_summary_df[
        [
            "preprocessing",
            "rule",
            "exact_accuracy",
            "almond_sensitivity",
            "peanut_sensitivity",
            "almond_specificity",
            "peanut_specificity",
            "unknown_rate",
            "ambiguous_rate",
        ]
    ].sort_values(
        ["peanut_sensitivity", "peanut_specificity", "unknown_rate"],
        ascending=[False, False, True],
    )
)
```

```python
comparison_plot_df = simca_summary_df.copy()
comparison_plot_df["config"] = (
    comparison_plot_df["preprocessing"]
    + " | "
    + comparison_plot_df["rule"]
)

plot_bar_values(
    comparison_plot_df["config"],
    comparison_plot_df["peanut_sensitivity"],
    title="SIMCA comparison — peanut sensitivity",
    x_title="Preprocessing | rule",
    y_title="Peanut sensitivity",
)

plot_bar_values(
    comparison_plot_df["config"],
    comparison_plot_df["unknown_rate"],
    title="SIMCA comparison — unknown rate",
    x_title="Preprocessing | rule",
    y_title="Unknown rate",
)

plot_bar_values(
    comparison_plot_df["config"],
    comparison_plot_df["ambiguous_rate"],
    title="SIMCA comparison — ambiguous rate",
    x_title="Preprocessing | rule",
    y_title="Ambiguous rate",
)
```

```python
simca_summary_df["selection_score"] = (
    simca_summary_df["peanut_sensitivity"]
    + simca_summary_df["peanut_specificity"]
    - simca_summary_df["unknown_rate"]
    - simca_summary_df["ambiguous_rate"]
)

best_simca_configs = (
    simca_summary_df
    .sort_values("selection_score", ascending=False)
    .head(5)
)

display(best_simca_configs[
    [
        "preprocessing",
        "rule",
        "selection_score",
        "exact_accuracy",
        "almond_sensitivity",
        "peanut_sensitivity",
        "almond_specificity",
        "peanut_specificity",
        "unknown_rate",
        "ambiguous_rate",
    ]
])
```
