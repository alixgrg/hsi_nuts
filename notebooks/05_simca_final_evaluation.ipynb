# 05 — SIMCA final evaluation

This notebook evaluates the selected SIMCA configurations on the reserved final test set.

Experimental protocol:

- Hyperparameters were selected in notebook `04_simca_model_selection.ipynb`.
- The final test batch was not used during model selection.
- Final training uses pure peanut objects from batches 1, 2 and 3.
- Final evaluation uses pure almond and pure peanut objects from batch 4.
- Mixture images are still not used here.

Main outputs:

- final object-level metrics;
- final metrics by batch and by image;
- final object-level predictions;
- final pixel-level error summaries;
- optional border/core sensitivity analysis;
- optional three-way decision analysis.

All result tables are saved as `.parquet`.

```python
def count_objects_matching_filters(meta_df: pd.DataFrame, filters: dict) -> pd.DataFrame:
    df = meta_df.copy()
    mask = pd.Series(True, index=df.index)

    for col, allowed in filters.items():
        if col not in df.columns:
            raise KeyError(f"Missing column in object metadata: {col}")

        if allowed is None:
            continue

        mask &= df[col].isin(list(allowed))

    return df[mask].copy()


final_train_meta_df = count_objects_matching_filters(object_meta_df, FINAL_TRAIN_FILTERS)
final_test_meta_df = count_objects_matching_filters(object_meta_df, FINAL_TEST_FILTERS)
```

```python
def _get_value(row, col, default):
    if col not in row.index:
        return default
    value = row[col]
    if pd.isna(value):
        return default
    return value


def normalize_selected_configs(selected_configs_df: pd.DataFrame) -> pd.DataFrame:
    rows = []

    for i, row in selected_configs_df.iterrows():
        model_family = str(_get_value(row, "model_family", "standard_rule"))
        selection_profile = str(_get_value(row, "selection_profile", f"config_{i:02d}"))

        preprocessing = str(row["preprocessing"])

        preprocessing_steps = _get_value(row, "preprocessing_steps", preprocessing)
        if isinstance(preprocessing_steps, str):
            preprocessing_steps_tuple = tuple(preprocessing_steps.split("+"))
        else:
            preprocessing_steps_tuple = tuple(preprocessing_steps)

        rule = str(_get_value(row, "rule", _get_value(row, "selected_rule_name", "alternative")))
        rule_variant = str(_get_value(row, "rule_variant", _get_value(row, "selected_rule_name", rule)))

        matrix_method = str(_get_value(row, "matrix_method", DEFAULT_MATRIX_METHOD))
        balanced_pixel_strategy = str(
            _get_value(row, "balanced_pixel_strategy", DEFAULT_BALANCED_PIXEL_STRATEGY)
        )

        m = _get_value(row, "m", DEFAULT_M)
        if pd.isna(m):
            m = DEFAULT_M

        out = {
            "config_id": f"{i:02d}_{selection_profile}",
            "selection_profile": selection_profile,
            "model_family": model_family,
            "matrix_method": matrix_method,
            "balanced_pixel_strategy": balanced_pixel_strategy,
            "preprocessing": preprocessing,
            "preprocessing_steps": "+".join(preprocessing_steps_tuple),
            "rule": rule,
            "rule_variant": rule_variant,
            "selected_rule_name": str(_get_value(row, "selected_rule_name", rule_variant)),
            "n_components": int(_get_value(row, "n_components", 5)),
            "alpha": float(_get_value(row, "alpha", DEFAULT_ALPHA)),
            "object_threshold": float(_get_value(row, "object_threshold", DEFAULT_OBJECT_THRESHOLD)),
            "m": int(m),
            "sg_window_length": int(_get_value(row, "sg_window_length", DEFAULT_SG_WINDOW_LENGTH)),
            "sg_polyorder": int(_get_value(row, "sg_polyorder", DEFAULT_SG_POLYORDER)),
            "position_dilation_radius": int(
                _get_value(row, "position_dilation_radius", DEFAULT_POSITION_DILATION_RADIUS)
            ),
        }

        rows.append(out)

    out_df = pd.DataFrame(rows)

    out_df = out_df.drop_duplicates(
        subset=[
            "model_family",
            "matrix_method",
            "balanced_pixel_strategy",
            "preprocessing",
            "preprocessing_steps",
            "rule",
            "rule_variant",
            "n_components",
            "alpha",
            "object_threshold",
            "m",
            "sg_window_length",
            "sg_polyorder",
            "position_dilation_radius",
        ]
    ).reset_index(drop=True)

    return out_df


final_configs_df = normalize_selected_configs(selected_configs_df)

display(final_configs_df)
```

```python
def add_final_selection_score(df: pd.DataFrame) -> pd.DataFrame:
    """
    Same orientation as notebook 04:
    minimize FN first, then FP, then maximize F1/accuracy.
    """
    out = df.copy()

    if "fn_rate" not in out.columns:
        out["fn_rate"] = 1.0 - out["peanut_sensitivity"].astype(float)

    if "fp_rate" not in out.columns:
        out["fp_rate"] = 1.0 - out["almond_specificity"].astype(float)

    out["final_score"] = (
        -10.0 * out["fn_rate"].astype(float)
        -1.0 * out["fp_rate"].astype(float)
        +0.02 * out.get("f1_score", 0.0)
        +0.05 * out.get("accuracy", 0.0)
        +0.02 * out.get("balanced_accuracy", 0.0)
    )

    return out


def sort_final_metrics(df: pd.DataFrame) -> pd.DataFrame:
    if df.empty:
        return df.copy()

    df = add_final_selection_score(df)

    sort_cols = [
        "fn_rate",
        "fp_rate",
        "accuracy",
        "f1_score",
        "balanced_accuracy",
        "final_score",
    ]

    sort_cols = [c for c in sort_cols if c in df.columns]

    ascending = [
        True if c in ["fn_rate", "fp_rate"] else False
        for c in sort_cols
    ]

    return df.sort_values(sort_cols, ascending=ascending).reset_index(drop=True)
```

```python
def run_standard_final_evaluation(row: pd.Series):
    steps = tuple(str(row["preprocessing_steps"]).split("+"))

    res = run_single_simca_pixel_projection(
        object_db=object_db,
        image_db=image_db,
        matrix_method=str(row["matrix_method"]),
        preprocessing_name=str(row["preprocessing"]),
        preprocessing_steps=steps,
        rule_name=str(row["rule"]),
        train_filters=FINAL_TRAIN_FILTERS,
        projection_filters=FINAL_TEST_FILTERS,
        object_thresholds=[float(row["object_threshold"])],
        n_components=int(row["n_components"]),
        alpha=float(row["alpha"]),
        m=int(row["m"]),
        random_state=RANDOM_STATE,
        replace=REPLACE_BALANCED_PIXELS,
        wavelengths=wavelengths,
        sg_window_length=int(row["sg_window_length"]),
        sg_polyorder=int(row["sg_polyorder"]),
        position_dilation_radius=int(row["position_dilation_radius"]),
        balanced_pixel_strategy=str(row["balanced_pixel_strategy"]),
    )

    object_threshold = float(row["object_threshold"])
    obj_df = res["object_tables"][object_threshold]
    pix_df = res["pixel_df"].copy()

    return {
        "result_type": "standard_rule",
        "pixel_df": pix_df,
        "object_df": obj_df,
        "threshold_df": res["threshold_df"],
    }


def run_empirical_final_evaluation(row: pd.Series):
    steps = tuple(str(row["preprocessing_steps"]).split("+"))
    preprocessing_name = str(row["preprocessing"])
    rule_variant = str(row["rule_variant"])

    single_preproc = {
        preprocessing_name: steps
    }

    summary_df, results, errors_df = run_simca_empirical_rule_grid(
        object_db=object_db,
        image_db=image_db,
        train_filters=FINAL_TRAIN_FILTERS,
        projection_filters=FINAL_TEST_FILTERS,
        preprocessing_configs=single_preproc,
        rule_variants=[rule_variant],
        n_components_values=[int(row["n_components"])],
        matrix_method=str(row["matrix_method"]),
        alpha=float(row["alpha"]),
        object_threshold=float(row["object_threshold"]),
        m=int(row["m"]),
        random_state=RANDOM_STATE,
        replace=REPLACE_BALANCED_PIXELS,
        wavelengths=wavelengths,
        sg_window_length=int(row["sg_window_length"]),
        sg_polyorder=int(row["sg_polyorder"]),
        position_dilation_radius=int(row["position_dilation_radius"]),
        cv_n_splits=CV_N_SPLITS,
        group_col=CV_GROUP_COL,
        keep_pixel_tables=True,
        keep_cv_tables=False,
        verbose=True,
        balanced_pixel_strategy=str(row["balanced_pixel_strategy"]),
    )

    if errors_df is not None and len(errors_df) > 0:
        display(errors_df)

    if len(results) == 0:
        raise RuntimeError("Empirical final evaluation produced no stored result.")

    first_key = next(iter(results.keys()))
    stored = results[first_key]

    obj_df = stored["object_tables_by_rule"][rule_variant].copy()
    pix_df = stored["pixel_variants_df"].copy()

    pix_df["predicted_peanut_pixel"] = pix_df[f"pred_{rule_variant}"].astype(bool)
    pix_df["rule_statistic"] = pix_df[f"stat_{rule_variant}"]
    pix_df["rule_limit"] = pix_df[f"limit_{rule_variant}"]
    pix_df["rule_name"] = rule_variant

    return {
        "result_type": "empirical_cv_rule",
        "pixel_df": pix_df,
        "object_df": obj_df,
        "threshold_df": summary_df,
        "errors_df": errors_df,
        "cv_thresholds": stored.get("cv_thresholds"),
        "cv_calibration_summary": stored.get("cv_calibration_summary"),
    }


def run_final_evaluation_for_config(row: pd.Series):
    model_family = str(row["model_family"])

    if model_family == "standard_rule":
        return run_standard_final_evaluation(row)

    if model_family == "empirical_cv_rule":
        return run_empirical_final_evaluation(row)

    # Fallback: try standard workflow.
    print(f"[WARNING] Unknown model_family={model_family!r}. Trying standard workflow.")
    return run_standard_final_evaluation(row)
```

```python
metrics_by_batch_parts = []
metrics_by_image_parts = []

for config_id, obj_df in final_object_tables.items():
    if "batch" in obj_df.columns:
        batch_df = metrics_by_group(
            obj_df,
            group_col="batch",
            true_col="true_peanut_object",
            pred_col="predicted_peanut_object",
        )
        batch_df["config_id"] = config_id
        metrics_by_batch_parts.append(batch_df)

    if "source_image" in obj_df.columns:
        image_df = metrics_by_group(
            obj_df,
            group_col="source_image",
            true_col="true_peanut_object",
            pred_col="predicted_peanut_object",
        )
        image_df["config_id"] = config_id
        metrics_by_image_parts.append(image_df)

final_metrics_by_batch_df = (
    pd.concat(metrics_by_batch_parts, ignore_index=True, sort=False)
    if metrics_by_batch_parts
    else pd.DataFrame()
)

final_metrics_by_image_df = (
    pd.concat(metrics_by_image_parts, ignore_index=True, sort=False)
    if metrics_by_image_parts
    else pd.DataFrame()
)

display(final_metrics_by_batch_df)
display(final_metrics_by_image_df)

save_parquet(final_metrics_by_batch_df, FINAL_METRICS_BY_BATCH_PATH)
save_parquet(final_metrics_by_image_df, FINAL_METRICS_BY_IMAGE_PATH)

print("Saved:")
print(" -", FINAL_METRICS_BY_BATCH_PATH)
print(" -", FINAL_METRICS_BY_IMAGE_PATH)
```

```python
final_object_error_parts = []

for config_id, obj_df in final_object_tables.items():
    d = obj_df.dropna(subset=["true_peanut_object", "predicted_peanut_object"]).copy()

    errors = d[
        d["true_peanut_object"].astype(bool)
        != d["predicted_peanut_object"].astype(bool)
    ].copy()

    if len(errors) == 0:
        continue

    errors["config_id"] = config_id
    final_object_error_parts.append(errors)

final_object_errors_df = (
    pd.concat(final_object_error_parts, ignore_index=True, sort=False)
    if final_object_error_parts
    else pd.DataFrame()
)

display(final_object_errors_df)

FINAL_OBJECT_ERRORS_PATH = RESULTS_DIR / "final_object_errors.parquet"
save_parquet(final_object_errors_df, FINAL_OBJECT_ERRORS_PATH)

print("Saved final object errors:")
print(FINAL_OBJECT_ERRORS_PATH)
```

```python
three_way_parts = []
three_way_summary_parts = []

for config_id, obj_df in final_object_tables.items():
    config_row = final_configs_df[
        final_configs_df["config_id"].eq(config_id)
    ].iloc[0]

    upper_threshold = float(config_row["object_threshold"])

    obj_3way = add_three_way_object_decision(
        obj_df,
        target_class=TARGET_CLASS,
        lower_threshold=THREE_WAY_LOWER_THRESHOLD,
        upper_threshold=upper_threshold,
    )

    obj_3way["config_id"] = config_id
    three_way_parts.append(obj_3way)

    summary = summarize_three_way_decision(obj_3way)
    summary["config_id"] = config_id
    summary["lower_threshold"] = THREE_WAY_LOWER_THRESHOLD
    summary["upper_threshold"] = upper_threshold

    if "true_label_object" in obj_3way.columns:
        cross = (
            obj_3way
            .groupby(["decision_3way", "true_label_object"], dropna=False)
            .size()
            .reset_index(name="n")
        )
        cross["config_id"] = config_id
        cross["lower_threshold"] = THREE_WAY_LOWER_THRESHOLD
        cross["upper_threshold"] = upper_threshold
        three_way_summary_parts.append(cross)
    else:
        three_way_summary_parts.append(summary)

three_way_objects_df = (
    pd.concat(three_way_parts, ignore_index=True, sort=False)
    if three_way_parts
    else pd.DataFrame()
)

three_way_summary_df = (
    pd.concat(three_way_summary_parts, ignore_index=True, sort=False)
    if three_way_summary_parts
    else pd.DataFrame()
)

display(three_way_summary_df)

save_parquet(three_way_summary_df, FINAL_THREE_WAY_SUMMARY_PATH)
save_parquet(three_way_objects_df, RESULTS_DIR / "final_three_way_object_tables.parquet")

print("Saved three-way analysis:")
print(" -", FINAL_THREE_WAY_SUMMARY_PATH)
```

```python
def list_result_files(directory):
    rows = []

    for path in Path(directory).rglob("*"):
        if path.is_file():
            rows.append({
                "file": str(path.relative_to(directory)),
                "suffixes": "".join(path.suffixes),
                "size_mb": path.stat().st_size / 1024**2,
            })

    if not rows:
        return pd.DataFrame(columns=["file", "suffixes", "size_mb"])

    return (
        pd.DataFrame(rows)
        .sort_values("size_mb", ascending=False)
        .reset_index(drop=True)
    )


files_df = list_result_files(RESULTS_DIR)
display(files_df)

forbidden = files_df[
    ~files_df["suffixes"].eq(".parquet")
]

if len(forbidden) > 0:
    print("[WARNING] Non-parquet result files found:")
    display(forbidden)
else:
    print("[OK] All final evaluation result files are parquet.")
```
